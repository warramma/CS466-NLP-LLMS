# Intro to working with LLMS

```python
import os
from dotenv import load_dotenv
load_dotenv()
os.environ["GROQ_API_KEY"]== os.getenv("GROQ_API_KEY")
```

```python
from langchain.chat_models import init_chat_model
model = init_chat_model(model="openai/gpt-oss-120B", model_provider="groq")
model
```

```python
response = model.invoke("greet me in casual Malay")
```

```python
response.content
```

```python
model = init_chat_model(model="gemini-3.6-flash", model_provider="google_genai")
model
```

```python
response = model.invoke("greet me in casual Malay").content
```
